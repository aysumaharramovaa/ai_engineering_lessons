```python
#Celsius ilə temperaturlar list-i verilib: `celsius_temps = [0, 10, 20, 30, 40]`.
#2.  Hər temperaturu Fahrenheit-ə çevirmək üçün `map()` function-ından `lambda` ilə istifadə edin. Formula: `(C * 9/5) + 32`.
#3.  Map obyektini list-ə çevirin və çap edin.
#4.  Eyni nəticəni daha qısa yolla əldə etmək üçün list comprehension istifadə edin.

celsius_temps = [0, 10, 20, 30, 40]
f_temps = map(
    lambda c: (c * 9/5) + 32,
    celsius_temps
)
print(list(f_temps))
```

```python
# Using `filter`
#1.  Rəqəmlər list-i verilib: `numbers = [1, 15, 22, 37, 44, 59, 60, 72]`.
#2.  Həm 2-yə, həm də 3-ə bölünən rəqəmləri əldə etmək üçün `filter()` function-ından `lambda` ilə istifadə edin.
#3.  Filter obyektini list-ə çevirin və çap edin.
#4.  Eyni nəticəni daha qısa yolla əldə etmək üçün list comprehension istifadə edin.

numbers = [1, 15, 22, 37, 44, 59, 60, 72]
bolunme = filter(lambda x: x%6==0, numbers)
print(list(bolunme))
```

```python
#Using `reduce`
#1.  Rəqəmlər list-i verilib: `numbers = [1, 2, 3, 4, 5]`.
#2.  `functools` module-undan `reduce`-u import edin.
#3.  List-dəki bütün rəqəmlərin cəmini tapmaq üçün `reduce`-dan `lambda` function ilə istifadə edin.
#4.  Nəticəni çap edin.

from functools import reduce
numbers = [1, 2, 3, 4, 5]
cem = reduce(lambda x,y : x+y, numbers)
print(cem)
```

```python
# List Comprehension
#1.  1-dən 10-a qədər rəqəmlərin kublarından ibarət list yaratmaq üçün list comprehension istifadə edin.

kub = [i**3 for i in range(1,11)]
print(kub)
```

```python
#List Comprehension with Condition
#1.  Sözlər list-i verilib: `words = ["Programming", "is", "fun", "and", "powerful"]`.
#2.  Yalnız 5-dən çox simvola malik sözləri ehtiva edən yeni list yaratmaq üçün list comprehension istifadə edin.

words = ["Programming", "is", "fun", "and", "powerful"]
long_words = [word for word in words if len(word) > 5]
print(long_words)
```

```python
#Dictionary Comprehension
#1.  Key-ləri 1-dən 5-ə qədər rəqəmlər, value-ları isə onların müvafiq faktorialları olan dictionary yaradın (məs., `5! = 1*2*3*4*5`). Faktorial üçün kiçik köməkçi (helper) function yazmaq istəyə bilərsiniz.
#    *İpucu (Hint): Function çağıran dictionary comprehension yarada bilərsiniz.

def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

dc = {
    x: factorial(x)
    for x in range(1, 6)
}

print(dc)
```

```python
# Set Comprehension
#1.  Təkrarlanan rəqəmləri olan list verilib: `numbers = [1, 2, 3, 2, 4, 5, 4, 1]`.
#2.  List-dəki unikal rəqəmlərin kvadratlarından ibarət set yaratmaq üçün set comprehension istifadə edin.

numbers = [1, 2, 3, 2, 4, 5, 4, 1]
result = {
    x**2 for x in numbers
}

print(result)
```

```python
#Basic `try...except`
#1.  İstifadəçidən index daxil etməsini soruşan proqram yazın.
#2.  `my_list = [10, 20, 30, 40]` list-indən istifadə edin.
#3.  İstifadəçi yanlış index daxil etdikdə baş verə biləcək `IndexError`-u idarə etmək üçün `try...except` blokundan istifadə edin. Səmimi (friendly) xəta mesajı çap edin.
# 4.  Həmçinin, istifadəçi rəqəm olmayan nəsə daxil edərsə, `ValueError`-u idarə edin.

my_list = [10, 20, 30, 40]
try:
    index = int(input("Index daxil edin: "))
    print(my_list[index])
except IndexError:
    print("xəta belə bir index mövcud deyil. Yenidən cəhd edin.")
except ValueError:
    print("xəta zəhmət olmasa, yalnız rəqəm daxil edin.")
```

```python
#12
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error! Cannot divide by zero.")
    else:
        print("Result:", result)
    finally:
        print("Division attempt finished.")
divide(10, 2)
divide(10, 0)
```

```python
#13
def check_password_strength(password):
    if len(password) < 8:
        raise ValueError("Password is too short!")
    else:
        print("Password is strong.")
try:
    check_password_strength("abc")
except ValueError as e:
    print(e)
try:
    check_password_strength("password123")
except ValueError as e:
    print(e)
```

```python
#14
def calculate_discount(price, discount_percent):
    assert 0 <= discount_percent <= 100, "Invalid discount percentage."
    final_price = price - (price * discount_percent / 100)
    return final_price
try:
    print(calculate_discount(100, 20))
except AssertionError as e:
    print(e)
try:
    print(calculate_discount(100, 110))
except AssertionError as e:
    print(e)
```
