```python
#restoran sifariş sistemi
menyu = {
    "pizza": 25,
    "burger": 12,
    "salat": 8,
    "dondurma": 18,
    "spagetti": 45,
    "hatay_doner": 20,
}
sifaris = ["hatay_doner", "burger", "pizza"]
total = 0
for yemek in sifaris:
    qiymet = menyu[yemek]
    print(f"{yemek} - {qiymet} azn")
    total += qiymet
print(f"umumi mebleg: {total} azn")
if total > 100:
    endirim = 20
elif total > 50:
    endirim = 10
else:
    endirim = 0
endirim_meblegi = total * endirim / 100
yekun = total - endirim_meblegi

print(f"endirim faizi: {endirim}%")
print(f"endirim mebleg: {endirim_meblegi} azn")
print(f"yekun odenilek: {yekun} azn")

```

```python
def factorial(n):
    fact = 1
    for num in range(2, n + 1):
        fact *= num
    return fact

print(factorial(3))
```

```python
#Verilmiş intervalda yerləşən bütün sadə ədədləri tapan funksiya yaz.
def sade_ededler(ilk, son):
    netice = []
    for eded in range(ilk, son + 1):
        if eded < 2:
            continue
        sade = True
        for bolen in range(2, int(eded ** 0.5) + 1):
            if eded % bolen == 0:
                sade = False
                break
        if sade:
            netice.append(eded)
    return netice
print(sade_ededler(10, 30))  
```

```python
#Verilmiş cümlədəki ən uzun sözün uzunluğunu qaytaran funksiya yaz.
def en_uzun_soz_uzunlugu(cumle):
    sozler = cumle.split()
    if not sozler:
        return 0
    return max(len(soz) for soz in sozler)
print(en_uzun_soz_uzunlugu(input()))
```

```python
#Verilmiş listdəki təkrarlanan elementləri silərək yeni list qaytaran funksiya yaz.
def tekrarlari_sil(siyahi):
    yeni_siyahi = []
    for element in siyahi:
        if element not in yeni_siyahi:
            yeni_siyahi.append(element)
    return yeni_siyahi
print(tekrarlari_sil([1, 2, 2, 3, 1]))       
```

```python
#Verilmiş string-in palindrom olub-olmadığını yoxlayan funksiya yaz
def palindrom(metn):
    metn = ''.join(metn.lower().split())
    return metn == metn[::-1]
print(palindrom("ata"))    
print(palindrom("salam"))          
print(palindrom(input()))    
```

```python
#sozdeki sailtlerin sayini tapan proqram
def setirlerin_sayi(soz):
    saitler = "a,e,ə,i,o,u,ü,u".split(",")
    say = 0
    for herf in soz.lower():
        if herf in saitler:
            say += 1
    return say
print(setirlerin_sayi(input()))
```

```python
cars = {
    "BMW": {"price": 45000, "year": 2020},
    "Mercedes": {"price": 60000, "year": 2022},
    "Toyota": {"price": 25000, "year": 2018},
    "Kia": {"price": 30000, "year": 2021},
    "Audi": {"price": 50000, "year": 2019},
    "Hyundai": {"price": 28000, "year": 2023}
}


#Qiyməti 30 000 AZN-dən aşağı olan avtomobilləri tapın.
#qiymətə görə avtomobilləri artan sıra ilə sıralayın.
#Qiyməti 40 000 AZN-dən çox olan avtomobillərin sayını hesablayın.
#Verilmiş mətn daxilində hansı sözün daha çox təkrarlandığını müəyyən edən proqram yazın.
#text = "Python proqramlaşdırma dilidir. Python öyrənmək maraqlıdır.
#Proqramlaşdırma məntiq tələb edir. Python ilə proqramlaşdırma etmək mümkündür."

# 30 000 AZN-dən ucuz avtomobillər
for name in cars:
    if cars[name]["price"] < 30000:
        print(name, cars[name]["price"])

# Qiyməti artan sıra ilə avtomobillər
names = list(cars.keys())

for i in range(len(names)):
    for j in range(i + 1, len(names)):
        if cars[names[i]]["price"] > cars[names[j]]["price"]:
            names[i], names[j] = names[j], names[i]

for name in names:
    print(name, cars[name]["price"])

# 40 000 AZN-dən baha avtomobillərin sayı
count = 0

for name in cars:
    if cars[name]["price"] > 40000:
        count = count + 1

print("bahalı avtomobillərin sayı:", count)
```
