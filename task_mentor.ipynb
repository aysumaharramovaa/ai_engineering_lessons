```python
#Python dilində dictionary və şərt operatorlarından istifadə edərək sadə bank sistemi hazırlamalısan. Proqram istifadəçinin məlumatlarını yoxlamalı, hesabına daxil olmasına icazə verməli və pul çıxarma əməliyyatını həyata keçirməlidir.

accounts = {
    "Elton": {"password": "1234", "balance": 1500},
    "Ali": {"password": "5678", "balance": 800},
    "Aysel": {"password": "9999", "balance": 2500}
}
#1. İstifadəçinin yoxlanılması
#- İstifadəçidən adını daxil etməsini istə.
ad = input("ad")
#- Əgər daxil edilən ad accounts dictionary-sində yoxdursa, ekrana "İstifadəçi tapılmadı!" yazdır.
if ad not in accounts:
    print("bele istifadecimiz yoxdur")    
#- Əgər istifadəçi mövcuddursa, növbəti mərhələyə keç.
if ad in accounts:
#2. Şifrənin yoxlanılması
#- İstifadəçidən şifrəsini daxil etməsini istə.
    password = input("sifrenizi yazin")
#- Əgər şifrə yanlışdırsa, "Şifrə yanlışdır!" mesajını göstər.
if password != accounts[ad]["password"]:
    print("Şifrə yanlışdır!")
#- Əgər şifrə doğrudursa, istifadəçini salamla.
else:
    print("salam")
#4. Pul çıxarılması

#İstifadəçidən hesabından çıxarmaq istədiyi məbləği soruş.
pul=int(input("ne qewder pul lazimdir size"))
balans = accounts[ad]["balance"]
#Daxil edilən məbləğə uyğun olaraq aşağıdakı şərtləri yoxla:
#- Məbləğ 0-a bərabər və ya mənfidirsə, "Yanlış məbləğ!" mesajını göstər.
if pul <=0:
    print("yanlis mebleg")
#- Məbləğ balansdan çoxdursa, "Balansınızda kifayət qədər pul yoxdur!" mesajını göstər.
if pul > balans:
    print("balansinda bu qeder pul yoxdur")
#- İstifadəçi təsdiq edərsə, əməliyyatı həyata keçir.
if pul <= balans:
    print("pulu cixar")
    print(balans-pul, "qalan mebleginiz")
#- İstifadəçi təsdiq etməzsə, əməliyyatı ləğv et.
#- Məbləğ bütün şərtlərə uyğundursa, balansdan həmin məbləği çıx.


```

```python
#restoran sifariş sistemi
menyu = {
    "pizza": 25,
    "burger": 12,
    "salat": 8,
    "dondurma": 18,
    "spagetti": 45,
    "hatay_doner": 20,
}
sifaris = ["hatay_doner", "burger", "pizza"]
total = 0
for yemek in sifaris:
    qiymet = menyu[yemek]
    print(f"{yemek} - {qiymet} azn")
    total += qiymet
print(f"umumi mebleg: {total} azn")
if total > 100:
    endirim = 20
elif total > 50:
    endirim = 10
else:
    endirim = 0
endirim_meblegi = total * endirim / 100
yekun = total - endirim_meblegi

print(f"endirim faizi: {endirim}%")
print(f"endirim mebleg: {endirim_meblegi} azn")
print(f"yekun odenilek: {yekun} azn")

```

```python

Hava proqnozu analizatoru
Tapşırığın şərtləri:
1. Şəhərlərin adlarını və 7 günlük temperaturlarını saxlamaq üçün hava adlı dictionary yarat.
2. Hər şəhərin həftəlik temperaturlarını siyahı (list) şəklində saxla.
3. for dövründən istifadə edərək hər şəhərin temperaturlarını hesabla.
4. Hər şəhər üçün orta temperaturu tap və ekrana çıxar.
5. Həftənin ən isti gününü və temperaturunu müəyyən et.
6. Həftənin ən soyuq gününü və temperaturunu müəyyən et.
7. Temperaturun dərəcəsinə görə geyim tövsiyəsi ver:
   - 10°C-dən aşağı: Palto geyin.
   - 10°C–20°C: Gödəkçə geyin.
   - 20°C-dən yuxarı: Köynək geyin.
8. İstifadəçi istədiyi şəhərin adını daxil etdikdə həmin şəhərin hava məlumatları göstərilsin.
9. Əgər şəhər lüğətdə yoxdursa, "Bu şəhər haqqında məlumat yoxdur!" mesajı göstərilsin.
İstifadə ediləcək Python mövzuları: dict, list, for, if-elif-else, in, len(), sum(), max(), min().



hava = {
    "baki": [15, 18, 20, 12, 9, 22, 17],
    "seki": [10, 14, 19, 21, 16, 12, 18],
    "gence": [5, 8, 12, 15, 7, 10, 13],
    "xirdalan": [18, 20, 22, 19, 24, 21, 17]
}

for seher, temperaturlar in hava.items():
    print(seher)
    cem = sum(temperaturlar)
    orta = cem / len(temperaturlar)
    print("Orta temperatur:", round(orta, 2), "C")
    isti = max(temperaturlar)
    soyuq = min(temperaturlar)
    print("en isti temperatur:", isti, "C")
    print("en soyuq temperatur:", soyuq, "C")
seher = input("\nŞəhərin adını daxil edin: ")

if seher in hava:
    temperaturlar = hava[seher]
    print(seher)
    gunler = ["bazar ertəsi", "çərşənbə axşamı", "çərşənbə", "cümə axşamı", "cümə", "şənbə", "bazar"]

    for i in range(len(temperaturlar)):

        temperatur = temperaturlar[i]

        print(gunler[i], "-", temperatur, "C")

        # Geyim tövsiyəsi
        if temperatur < 10:
            print("Palto geyin.")
        elif temperatur <= 20:
            print("Gödəkçə geyin.")
        else:
            print("Köynək geyin.")

    # Orta temperatur
    orta = sum(temperaturlar) / len(temperaturlar)

    print("\nOrta temperatur:", round(orta, 2), "C")

    # Ən isti və soyuq günlərin indeksi
    isti = max(temperaturlar)
    soyuq = min(temperaturlar)

    isti_gun = temperaturlar.index(isti)
    soyuq_gun = temperaturlar.index(soyuq)

    print("Həftənin ən isti günü:", gunler[isti_gun], "-", isti, "C")
    print("Həftənin ən soyuq günü:", gunler[soyuq_gun], "-", soyuq, "C")

else:
    print("Bu şəhər haqqında məlumat yoxdur!")


```
