```python
s = "Programming is fun!"
# 4-cü index-də hansı simvol (character) var?
print(s[4])
# String-in son simvolu nədir?
print(s[-1])
# s[11:13] slicing-i hansı substring-i verir?
print(s[11:13])
# "gram" substring-ini necə əldə etmək olar?
print(s[8:12])
# Bütün string-i necə tərsinə çevirmək (reverse) olar?
print(s[::-1])
```

```python
# tweet = "  WOW, what a great game! #sports #python  " string-i verilib:
tweet = "  WOW, what a great game! #sports #python  "
# Bütün tweet-i lowercase-ə çevirin.
tweet = tweet.lower()
# Əvvəlki və sonrakı boşluqları (whitespace) silin.
tweet = tweet.strip()
# "#python"-u "#DataScience" ilə əvəz edin.
tweet = tweet.replace("#python", "#DataScience")
# Təmizlənmiş tweet-i sözlərdən ibarət list-ə split edin.
tweet_list = tweet.split()
```
