```python
#The `__init__` Method
#1.  Exercise 1-dəki `Car` class-ını dəyişdirin.
#2.  Hələlik `drive` və `stop` method-larını silin.
#3.  `make`, `model` və `year`-i argument kimi qəbul edən `__init__` method-u əlavə edin.
#4.  `__init__`-in içində `make`, `model` və `year` üçün instance attribute-lər yaradın.
#5.  "This is a 2020 Ford Mustang." kimi string return edən `describe()` method-u əlavə edin.
#6.  Yeni `Car` class-ınızın instance-ını yaradın və onun `describe()` method-unun nəticəsini çap edin.


class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
    def describe(self):
        return f"This is a {self.year} {self.make} {self.model}."
my_car = Car("Ford", "Mustang", 2020)
print(my_car.describe())
```
