```python
#Ən sevdiyiniz filmi təmsil edən movie adlı dictionary yaradın. O, "title", "director", "year" və "genre" key-lərini ehtiva etməlidir.
#"title" key-inin value-sunu çap edin.
#"rating" üçün yeni key-value pair əlavə edin (məs., 8.5).
#"year" value-sunu yeni dəyərlə dəyişin.
#"genre" key-ini dictionary-dən silin.
#Yekun movie dictionary-sini çap edin.

movie = {
    "title": "Marvel",
    "director": "Stan Lee",
    "year": 2021,
    "genre": "Action"
}
print(movie["title"])
movie["rating"] = 8.5
print(movie)
movie["year"] = 2022
print(movie)
del movie["genre"]
print(movie)
```

```python
#Exercise 1-dəki movie dictionary-sindən istifadə edərək bütün key-ləri çap edin.
#Bütün value-ları çap edin.
#Bütün key-value pair-ləri tuple-lar şəklində çap edin.
#.get() method-undan istifadə edərək "producer" key-inə daxil olmağa çalışın. Key mövcud deyilsə, default value olaraq "Unknown" verin. Nəticəni çap edin.
#"director" key-inin dictionary-də olub-olmadığını yoxlayın və nəticəni çap edin (True və ya False).

movie = {
    "title": "Marvel",
    "director": "Stan Lee",
    "year": 2021,
    "genre": "Action"
}
print(movie.keys())
print(movie.values())
print(movie.items())
print(movie.get("producer", "Unknown"))
print("director" in movie)
```

```python
#Təkrarlanan rəqəmləri olan rəqəmlər list-i yaradın: numbers = [10, 20, 30, 20, 40, 10, 50].
#Yalnız unikal rəqəmləri əldə etmək üçün bu list-dən set yaradın. Set-i çap edin.
#Set-ə 60 rəqəmini əlavə edin.
#10 rəqəmini set-ə yenidən əlavə etməyə çalışın. Nə baş verir? (Konseptual sual)
#Yekun set-i çap edin.

numbers = [10, 20, 30, 20, 40, 10, 50]
unikal_numbers = set(numbers)
print(unikal_numbers)
unikal_numbers.add(60)
print(unikal_numbers)
unikal_numbers.add(10)
print(unikal_numbers)
```

```python
#Set Operations
#Aşağıdakı iki set verilib:

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

#İki set-in union-unu tapın və çap edin.
print(set1.union())
print(set2.union())
#İki set-in intersection-ını tapın və çap edin.
print(set1.intersection())
print(set2.intersection())
#set1 və set2 arasındakı difference-i (set1 - set2) tapın və çap edin.
print(set1.difference(set2))
#set2 və set1 arasındakı difference-i (set2 - set1) tapın və çap edin.
print(set2.difference(set1))
#set2 və set1-in symmetric difference-ini tapın və çap edin. (method və formuldan istifadə edərək)
print(set1.symmetric_difference(set2))
```
