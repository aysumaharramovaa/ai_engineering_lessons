
### Arithmetic Operators
Used to perform mathematical operations.

| Operator | Description    | Example     | Result |
| :------- | :------------- | :---------- | :----- |
| `+`      | Addition       | `5 + 2`     | `7`    |
| `-`      | Subtraction    | `5 - 2`     | `3`    |
| `*`      | Multiplication | `5 * 2`     | `10`   |
| `/`      | Division       | `5 / 2`     | `2.5`  |
| `//`     | Floor Division | `5 // 2`    | `2`    |
| `%`      | Modulus        | `5 % 2`     | `1`    |
| `**`     | Exponentiation | `5 ** 2`    | `25`   |

### Comparison Operators
Used to compare two values. They return a Boolean (`True` or `False`).

| Operator | Description              | Example     | Result |
| :------- | :----------------------- | :---------- | :----- |
| `==`     | Equal to                 | `5 == 2`    | `False`|
| `!=`     | Not equal to             | `5 != 2`    | `True` |
| `>`      | Greater than             | `5 > 2`     | `True` |
| `<`      | Less than                | `5 < 2`     | `False`|
| `>=`     | Greater than or equal to | `5 >= 5`    | `True` |
| `<=`     | Less than or equal to    | `5 <= 2`    | `False`|

> **Note:** Equality is checked with `==`, not `=`.

### Summary of Basic Data Types

| Data Type | Description                   | Example              | Key Characteristics                                    |
| :-------- | :---------------------------- | :------------------- | :----------------------------------------------------- |
| `int`     | Integer numbers               | `-2, -1, 0, 1, 2, 3, 4, 5`           | Whole numbers, positive or negative, without decimals. |
| `float`   | Floating-point numbers        | `-1.25, -1.0, -0.5, 0.0, 0.5, 1.0, 1.25`      | Numbers with a decimal point.                          |
| `str`     | String (sequence of characters) | `'a', 'aa', 'aaa', 'Hello!', '11 cats'`     | Textual data, enclosed in single or double quotes.     |
| `bool`    | Boolean                       | `True, False`  | Represents `True` or `False`. Used in conditional logic. |
| `NoneType`| Absence of a value            | `None`         | Singleton `None`, falsy; compare with `is` (not `==`). |

### Assignment Operators
Used to assign values to variables.

| Operator | Example   | Equivalent to |
| :------- | :-------- | :------------ |
| `=`      | `x = 5`   | `x = 5`       |
| `+=`     | `x += 3`  | `x = x + 3`   |
| `-=`     | `x -= 3`  | `x = x - 3`   |
| `*=`     | `x *= 3`  | `x = x * 3`   |
| `/=`     | `x /= 3`  | `x = x / 3`   |
| `//=`    | `x //= 3` | `x = x // 3`  |
| `%=`     | `x %= 3`  | `x = x % 3`   |
| `**=`    | `x **= 3` | `x = x ** 3`  |
