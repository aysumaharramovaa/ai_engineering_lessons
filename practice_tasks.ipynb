```python
#fruits adlı list yarat. İçərisində 5 meyvə olsun.
#1. İlk elementi çap et.
#2. Sonuncu elementi negative indexing ilə çap et.
#3. İkinci və üçüncü elementləri slicing ilə əldə et.
#4. List-in uzunluğunu çap et.

fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print(fruits[0])
print(fruits[-1])
print(fruits[1:2])
print(len(fruits))
```

```python
colors = ["red", "green", "blue"]
#Sonuna "yellow" əlavə et.
#Index 1-ə "purple" əlavə et.
#"blue" elementini sil.
#"black" elementini əlavə et.
#Yekun list-i çap et.
colors.append("yellow")
print(colors)
colors.insert(1, "purple")
print(colors)
colors.remove("blue")
print(colors)
colors.append("black")
print(colors)
```

```python
numbers = [5, 2, 8, 2, 10, 2, 4]
#List-i artan sıra ilə düz.
#2 rəqəminin neçə dəfə olduğunu tap.
#10 rəqəminin index-ini tap.
#List-i tərsinə çevir.
#List-in kopyasını yarat.
numbers.sort()
print(numbers)
print(numbers.count(2))
print(numbers.index(10))
print(numbers[::-1])
print(numbers.copy())
print(numbers)
```

```python
#Number Checker
#İstifadəçidən rəqəm al.
#- Rəqəm müsbətdirsə "Positive" çap et.
#- Mənfidirsə "Negative" çap et.
#- Sıfırdırsa "Zero" çap et.

number=int(input("eded"))
if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")
```
